```python
# ============================================================
# CUSTOMER CHURN PREDICTION USING PSO-BASED FEATURE SELECTION
# ============================================================

# ============================================================
# IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import (
    train_test_split,
    cross_val_score
)

from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    roc_curve
)

from sklearn.linear_model import LogisticRegression

from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier
)

from sklearn.neural_network import MLPClassifier

from xgboost import XGBClassifier

from imblearn.over_sampling import SMOTE

import pyswarms as ps


# ============================================================
# CONFIGURATION
# ============================================================

RANDOM_STATE = 42
TEST_SIZE = 0.2


# ============================================================
# LOAD DATASET
# ============================================================

print("\n================================================")
print("LOADING DATASET")
print("================================================")

df = pd.read_csv("../data/Telco_Churn_Dataset.csv")

print("\nDATASET SHAPE")
print(df.shape)

print("\nFIRST 5 ROWS")
print(df.head())

print("\nDATASET INFO")
print(df.info())

print("\nMISSING VALUES")
print(df.isnull().sum())


# ============================================================
# DATA CLEANING
# ============================================================

print("\n================================================")
print("DATA CLEANING")
print("================================================")

if "TotalCharges" in df.columns:

    df["TotalCharges"] = pd.to_numeric(
        df["TotalCharges"],
        errors="coerce"
    )

    df["TotalCharges"].fillna(
        df["TotalCharges"].median(),
        inplace=True
    )


# ============================================================
# FEATURE ENGINEERING
# ============================================================

print("\n================================================")
print("FEATURE ENGINEERING")
print("================================================")

# Average Monthly Spend

if all(col in df.columns for col in ["TotalCharges", "tenure"]):

    df["Avg_Monthly_Spend"] = (
        df["TotalCharges"] /
        (df["tenure"] + 1)
    )

# Remaining Customer Value

if all(col in df.columns for col in ["MonthlyCharges", "tenure"]):

    df["Remaining_Value"] = (
        df["MonthlyCharges"] *
        (72 - df["tenure"])
    )

# Total Services Used

service_columns = [

    "PhoneService",
    "MultipleLines",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

available_services = [

    col for col in service_columns
    if col in df.columns
]

if len(available_services) > 0:

    df["Service_Count"] = sum(

        (df[col] == "Yes").astype(int)

        for col in available_services
    )

# Support Services

support_columns = [

    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport"
]

available_support = [

    col for col in support_columns
    if col in df.columns
]

if len(available_support) > 0:

    df["Support_Services"] = sum(

        (df[col] == "Yes").astype(int)

        for col in available_support
    )

# Month-to-Month Contract

if "Contract" in df.columns:

    df["Is_Month_to_Month"] = (

        df["Contract"] == "Month-to-month"

    ).astype(int)

# Electronic Check

if "PaymentMethod" in df.columns:

    df["Is_Electronic_Check"] = (

        df["PaymentMethod"] == "Electronic check"

    ).astype(int)

# Fiber Optic User

if "InternetService" in df.columns:

    df["Is_Fiber_Optic"] = (

        df["InternetService"] == "Fiber optic"

    ).astype(int)

# Has Internet

if "InternetService" in df.columns:

    df["Has_Internet"] = (

        df["InternetService"] != "No"

    ).astype(int)

# Paperless Billing

if "PaperlessBilling" in df.columns:

    df["Uses_Paperless_Billing"] = (

        df["PaperlessBilling"] == "Yes"

    ).astype(int)


# ============================================================
# DROP UNNECESSARY COLUMNS
# ============================================================

print("\n================================================")
print("DROPPING UNNECESSARY COLUMNS")
print("================================================")

columns_to_drop = [
    "customerID"
]

existing_columns = [

    col for col in columns_to_drop
    if col in df.columns
]

if len(existing_columns) > 0:

    df.drop(
        columns=existing_columns,
        inplace=True
    )


# ============================================================
# TARGET ENCODING
# ============================================================

print("\n================================================")
print("TARGET ENCODING")
print("================================================")

df["Churn"] = df["Churn"].map({

    "Yes": 1,
    "No": 0
})

y = df["Churn"]

X = df.drop(
    "Churn",
    axis=1
)


# ============================================================
# TARGET DISTRIBUTION
# ============================================================

plt.figure(figsize=(6, 5))

sns.countplot(x=y)

plt.title("Target Variable Distribution")

plt.xlabel("Churn")
plt.ylabel("Count")

plt.savefig("../results/target_distribution.png")

plt.show()


# ============================================================
# ONE HOT ENCODING
# ============================================================

print("\n================================================")
print("ONE HOT ENCODING")
print("================================================")

categorical_columns = X.select_dtypes(
    include=['object']
).columns

X = pd.get_dummies(

    X,

    columns=categorical_columns,

    drop_first=True
)

print("\nFEATURE MATRIX SHAPE")
print(X.shape)


# ============================================================
# FEATURE SCALING
# ============================================================

print("\n================================================")
print("FEATURE SCALING")
print("================================================")

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)


# ============================================================
# TRAIN TEST SPLIT
# ============================================================

print("\n================================================")
print("TRAIN TEST SPLIT")
print("================================================")

X_train, X_test, y_train, y_test = train_test_split(

    X_scaled,
    y,

    test_size=TEST_SIZE,

    stratify=y,

    random_state=RANDOM_STATE
)

print("X_train Shape :", X_train.shape)
print("X_test Shape  :", X_test.shape)


# ============================================================
# APPLY SMOTE
# ============================================================

print("\n================================================")
print("APPLYING SMOTE")
print("================================================")

smote = SMOTE(
    random_state=RANDOM_STATE
)

X_train, y_train = smote.fit_resample(
    X_train,
    y_train
)

print("\nAFTER SMOTE")
print(pd.Series(y_train).value_counts())


# ============================================================
# PSO FEATURE SELECTION
# ============================================================

print("\n================================================")
print("PSO FEATURE SELECTION")
print("================================================")


# ============================================================
# FITNESS FUNCTION
# ============================================================

def fitness_function(particles):

    scores = []

    for particle in particles:

        selected_features = np.where(
            particle > 0.5
        )[0]

        if len(selected_features) == 0:

            scores.append(1)

            continue

        X_selected = X_train[:, selected_features]

        model = XGBClassifier(

            n_estimators=300,
            learning_rate=0.03,
            max_depth=6,
            subsample=0.85,
            colsample_bytree=0.85,
            eval_metric='logloss',
            random_state=RANDOM_STATE
        )

        cv_scores = cross_val_score(

            model,

            X_selected,

            y_train,

            cv=3,

            scoring='f1'
        )

        score = 1 - cv_scores.mean()

        scores.append(score)

    return np.array(scores)


# ============================================================
# PSO CONFIGURATION
# ============================================================

options = {

    'c1': 1.5,
    'c2': 1.5,
    'w': 0.7
}


# ============================================================
# INITIALIZE PSO
# ============================================================

optimizer = ps.single.GlobalBestPSO(

    n_particles=20,

    dimensions=X_train.shape[1],

    options=options
)


# ============================================================
# RUN PSO
# ============================================================

best_cost, best_position = optimizer.optimize(

    fitness_function,

    iters=30
)


# ============================================================
# SELECT BEST FEATURES
# ============================================================

selected_features = np.where(

    best_position > 0.5

)[0]


# ============================================================
# PRINT SELECTED FEATURES
# ============================================================

print("\n================================================")
print("SELECTED FEATURES")
print("================================================")

selected_feature_names = X.columns[selected_features]

print(selected_feature_names)

print("\nTOTAL SELECTED FEATURES :")

print(len(selected_feature_names))


# ============================================================
# CREATE OPTIMIZED DATASET
# ============================================================

X_train_opt = X_train[:, selected_features]

X_test_opt = X_test[:, selected_features]

print("\n================================================")
print("OPTIMIZED DATASET SHAPE")
print("================================================")

print("X_train_opt :", X_train_opt.shape)

print("X_test_opt  :", X_test_opt.shape)


# ============================================================
# FEATURE IMPORTANCE
# ============================================================

print("\n================================================")
print("FEATURE IMPORTANCE")
print("================================================")

rf_temp = RandomForestClassifier(

    n_estimators=500,

    max_depth=10,

    min_samples_leaf=4,

    max_features='sqrt',

    random_state=RANDOM_STATE,

    n_jobs=-1
)

rf_temp.fit(
    X_train_opt,
    y_train
)

feature_importance = pd.DataFrame({

    "Feature": selected_feature_names,

    "Importance": rf_temp.feature_importances_
})

feature_importance = feature_importance.sort_values(

    by="Importance",

    ascending=False
)

top10_features = feature_importance.head(10)

print(top10_features)


# ============================================================
# TOP 10 FEATURE VISUALIZATION
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(

    x="Importance",

    y="Feature",

    data=top10_features
)

plt.title("Top 10 Important Features")

plt.xlabel("Importance")
plt.ylabel("Features")

plt.savefig("../results/feature_importance.png")

plt.show()


# ============================================================
# CREATE TOP 10 DATASET
# ============================================================

top10_feature_names = top10_features[
    "Feature"
].tolist()

top10_indices = [

    list(selected_feature_names).index(feature)

    for feature in top10_feature_names
]

X_train_top10 = X_train_opt[:, top10_indices]

X_test_top10 = X_test_opt[:, top10_indices]

print("\nTOP 10 DATASET SHAPE")

print("X_train_top10 :", X_train_top10.shape)

print("X_test_top10  :", X_test_top10.shape)


# ============================================================
# CORRELATION HEATMAP
# ============================================================

corr_df = pd.DataFrame(

    X_train_top10,

    columns=top10_feature_names
)

plt.figure(figsize=(10, 8))

sns.heatmap(

    corr_df.corr(),

    annot=True,

    cmap="coolwarm"
)

plt.title("Top 10 Feature Correlation Heatmap")

plt.savefig("../results/correlation_heatmap.png")

plt.show()


# ============================================================
# DEFINE MODELS
# ============================================================

models = {

    "PSO + Logistic Regression":

        LogisticRegression(
            max_iter=1000,
            random_state=RANDOM_STATE
        ),

    "PSO + Random Forest":

        RandomForestClassifier(
            n_estimators=500,
            max_depth=10,
            min_samples_leaf=4,
            max_features='sqrt',
            random_state=RANDOM_STATE,
            n_jobs=-1
        ),

    "PSO + Gradient Boosting":

        GradientBoostingClassifier(
            n_estimators=300,
            learning_rate=0.05,
            max_depth=4,
            random_state=RANDOM_STATE
        ),

    "PSO + XGBoost":

        XGBClassifier(
            n_estimators=1000,
            learning_rate=0.01,
            max_depth=6,
            subsample=0.9,
            colsample_bytree=0.9,
            gamma=0.1,
            reg_alpha=0.1,
            reg_lambda=1,
            min_child_weight=3,
            eval_metric='logloss',
            random_state=RANDOM_STATE
        ),

    "PSO + Neural Network":

        MLPClassifier(
            hidden_layer_sizes=(128, 64),
            activation='relu',
            solver='adam',
            max_iter=500,
            random_state=RANDOM_STATE
        )
}


# ============================================================
# MODEL TRAINING & EVALUATION
# ============================================================

print("\n================================================")
print("MODEL TRAINING & EVALUATION")
print("================================================")

results = []

for name, model in models.items():

    print("\n================================================")
    print(f"MODEL : {name}")
    print("================================================")

    model.fit(
        X_train_top10,
        y_train
    )

    y_pred = model.predict(
        X_test_top10
    )

    if hasattr(model, "predict_proba"):

        y_prob = model.predict_proba(
            X_test_top10
        )[:, 1]

    else:

        y_prob = y_pred

    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    precision = precision_score(
        y_test,
        y_pred
    )

    recall = recall_score(
        y_test,
        y_pred
    )

    f1 = f1_score(
        y_test,
        y_pred
    )

    roc_auc = roc_auc_score(
        y_test,
        y_prob
    )

    cv_scores = cross_val_score(

        model,

        X_train_top10,

        y_train,

        cv=5,

        scoring='f1'
    )

    cv_mean = cv_scores.mean()

    results.append([

        name,
        accuracy,
        precision,
        recall,
        f1,
        roc_auc,
        cv_mean
    ])

    print(f"Accuracy  : {accuracy:.4f}")
    print(f"Precision : {precision:.4f}")
    print(f"Recall    : {recall:.4f}")
    print(f"F1 Score  : {f1:.4f}")
    print(f"ROC-AUC   : {roc_auc:.4f}")
    print(f"CV F1 Mean: {cv_mean:.4f}")

    print("\nCLASSIFICATION REPORT\n")

    print(classification_report(
        y_test,
        y_pred
    ))

    # Confusion Matrix

    cm = confusion_matrix(
        y_test,
        y_pred
    )

    plt.figure(figsize=(5, 4))

    sns.heatmap(
        cm,
        annot=True,
        fmt='d',
        cmap='Blues'
    )

    plt.title(f"{name} Confusion Matrix")

    plt.xlabel("Predicted")
    plt.ylabel("Actual")

    plt.savefig(
        f"../results/{name}_confusion_matrix.png"
    )

    plt.show()

    # ROC Curve

    fpr, tpr, thresholds = roc_curve(
        y_test,
        y_prob
    )

    plt.figure(figsize=(6, 5))

    plt.plot(fpr, tpr)

    plt.plot([0, 1], [0, 1], '--')

    plt.title(f"{name} ROC Curve")

    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")

    plt.savefig(
        f"../results/{name}_roc_curve.png"
    )

    plt.show()


# ============================================================
# FINAL MODEL COMPARISON
# ============================================================

results_df = pd.DataFrame(

    results,

    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score",
        "ROC-AUC",
        "CV F1 Mean"
    ]
)

print("\n================================================")
print("FINAL MODEL COMPARISON")
print("================================================")

print(results_df)


# ============================================================
# RESULTS VISUALIZATION
# ============================================================

# Accuracy Plot

plt.figure(figsize=(10, 6))

sns.barplot(
    x="Model",
    y="Accuracy",
    data=results_df
)

plt.title("Model Accuracy Comparison")

plt.xticks(rotation=10)

plt.savefig("../results/model_accuracy.png")

plt.show()

# ROC-AUC Plot

plt.figure(figsize=(10, 6))

sns.barplot(
    x="Model",
    y="ROC-AUC",
    data=results_df
)

plt.title("ROC-AUC Comparison")

plt.xticks(rotation=10)

plt.savefig("../results/model_roc_auc.png")

plt.show()

# F1 Score Plot

plt.figure(figsize=(10, 6))

sns.barplot(
    x="Model",
    y="F1-Score",
    data=results_df
)

plt.title("F1-Score Comparison")

plt.xticks(rotation=10)

plt.savefig("../results/model_f1_score.png")

plt.show()


# ============================================================
# SAVE RESULTS
# ============================================================

results_df.to_csv(
    "../results/final_model_results.csv",
    index=False
)

print("\n================================================")
print("RESULTS SAVED SUCCESSFULLY")
print("================================================")


# ============================================================
# CONCLUSION
# ============================================================

print("\n================================================")
print("PROJECT COMPLETED SUCCESSFULLY")
print("================================================")

print(
    "\nPSO-based feature selection improved model "
    "performance and reduced irrelevant features "
    "for customer churn prediction."
)
```
